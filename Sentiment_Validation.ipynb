# VADER Sentiment Validation

This notebook validates VADER sentiment classifications against star-rating-derived ground truth labels across the complete review dataset (n = 13,183). It uses only the publicly available `All_Reviews_Dataset.xlsx` file: ground truth labels are derived from the `Rating` column, and VADER's predictions are taken from the `Sentiment_Category` column in the same row. No sampling, no external files, and no row-matching step are involved, so the results below are exactly reproducible.

## 1. Import Libraries

```python
import pandas as pd
from sklearn.metrics import classification_report, accuracy_score
```

## 2. Load the Dataset

```python
df = pd.read_excel("All_Reviews_Dataset.xlsx")
# columns: Review_ID, Destination, Province, Review_Date, Rating,
#          VADER_Compound_Score, Sentiment_Category
df.head()
```

## 3. Derive Ground Truth Labels from Star Rating

Ratings of 4-5 stars are treated as Positive, 3 stars as Neutral, and 1-2 stars as Negative.

```python
def rating_to_label(rating):
    if rating <= 2:
        return "Negative"
    if rating == 3:
        return "Neutral"
    return "Positive"

df["Ground_Truth_Label"] = df["Rating"].apply(rating_to_label)
df[["Rating", "Ground_Truth_Label", "Sentiment_Category"]].head()
```

## 4. Compute Validation Metrics

Overall agreement and per-class precision, recall, and F1-score, computed across the complete dataset.

```python
accuracy = accuracy_score(df["Ground_Truth_Label"], df["Sentiment_Category"])
print(f"Overall agreement: {accuracy:.4f}")
print()
print(classification_report(df["Ground_Truth_Label"], df["Sentiment_Category"], digits=2))
```

## 5. Expected Output

Running this notebook on `All_Reviews_Dataset.xlsx` reproduces the validation figures reported in the manuscript:

| Class | Precision | Recall | F1 | Support |
|---|---|---|---|---|
| Negative | 0.52 | 0.51 | 0.51 | 1,471 |
| Neutral | 0.26 | 0.04 | 0.07 | 1,801 |
| Positive | 0.83 | 0.95 | 0.89 | 9,911 |

Overall agreement = 78.0%; weighted average F1 = 0.73.
